# WHICH HOTEL HAVE HIGH CANCEL RATES

```python
plt.figure(figsize=(8,4))
ax1=sns.countplot(x='hotel',hue='is_canceled',data=df)
plt.legend(['canceled','not_canceled'])
plt.show()
```

City Hotels attract more bookings and cancellations due to business travel and last-minute behavior, while Resort Hotel bookings are more stable but lower in volume.

```python

resort_hotel=df[df['hotel'] == 'Resort Hotel']
resort_hotel['is_canceled'].value_counts(normalize=True)

```

```python
city_hotel=df[df['hotel'] == 'City Hotel']
city_hotel['is_canceled'].value_counts(normalize=True)

```

```python
df.columns
```

Effect of price on city and resort hotel

```python
resort_hotel=resort_hotel.groupby(['reservation_status_date'],as_index=False)['adr'].mean()
city_hotel=city_hotel.groupby(['reservation_status_date'],as_index=False)['adr'].mean()
```

```python
plt.figure(figsize=(30,8))
plt.plot(resort_hotel.index,resort_hotel['adr'], label='resort')
plt.plot(city_hotel.index,city_hotel['adr'], label='city')
plt.ylabel('adr')
plt.xlabel('index')
plt.legend()
plt.show()
```

Resort hotels show high volatility and higher overall ADR, influenced heavily by seasonal demand.
City hotels maintain stable, moderate pricing with gradual upward trends.

which month have higher reservations

```python
df.columns
```

```python

df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'], errors='coerce')
df['month']=df['reservation_status_date'].dt.month
plt.figure(figsize=(18,8))
ax1=sns.countplot(x='month',hue='is_canceled',data=df)
plt.xlabel('month')
plt.ylabel('no of reservation')
plt.title('reservation per month')
```

Summer months bring the most bookings with very low cancellations, while winter months (especially January and December) see high cancellation rates despite fewer bookings.
